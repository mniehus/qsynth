# 01 · Exact policy lookup for two regimes

Use the Python environment from [Notebook 00](00_environment_and_bell.ipynb). Each notebook in this sequence can be run from a fresh kernel.

## Question

Which policy should the information engine use in each operating regime?

## Concept

**Informatics view.** A regime is an input category and a policy is a candidate configuration. In the engine, the configuration sets how the estimator infers bead position and how the controller moves the trap.

A regime describes the operating conditions. A policy is a fixed set of estimator and controller parameters. The lookup table assigns one policy to each regime.

$$J_{ij}=\text{performance of policy }j\text{ in regime }i.$$

**Illustrative matrix, arbitrary performance units. Higher is better.** The scores are the negatives of the local costs in the combined reference notebook, used here with independent regime choices.

| Regime | π₁ · nominal | π₂ · robust |
|---|---:|---:|
| ρ₁ · baseline sensor noise | −1.0 | −3.0 |
| ρ₂ · elevated sensor noise | −2.5 | −1.0 |

For example, $J_{21}=-2.5$ scores the nominal policy under elevated noise. The first policy in a map applies to ρ₁, and the second to ρ₂. The four maps are $(\pi_1,\pi_1)$, $(\pi_1,\pi_2)$, $(\pi_2,\pi_1)$, and $(\pi_2,\pi_2)$. Add the two selected entries to score a map.

## Check / exercise

Predict the winning map from the table. Then predict which map wins if $J_{22}$ drops from −1.0 to −4.0.

## Worked example and result

```python
from itertools import product

J = [[-1.0, -3.0], [-2.5, -1.0]]
labels = ("π₁ nominal", "π₂ robust")
maps = list(product(range(2), repeat=2))
scores = {pair: sum(J[i][j] for i, j in enumerate(pair)) for pair in maps}

for (first, second), score in scores.items():
    print(f"{labels[first]:12} | {labels[second]:12} | {score:5.1f}")

best_score = max(scores.values())
winners = [pair for pair, score in scores.items() if score == best_score]
assert winners == [(0, 1)] and best_score == -2.0
print("Exact optimum:", tuple(labels[j] for j in winners[0]), best_score)
```

**Exercise check.** If the robust score in ρ₂ falls to −4.0, nominal wins in both regimes, with a total of −3.5.

## Takeaway

The initial synthesis problem is a small exact lookup table. With additive scores, selecting the best policy in each row gives the best global map.
