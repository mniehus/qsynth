# Offline optimization and policy synthesis

A local score compares one candidate policy in one regime. A regime graph and switching penalty define dependencies between choices. The synthesized object is a map from regimes to policies that can be deployed in the classical loop.

**Established.** [Exact lookup](01_policy_lookup_2x2.ipynb), [simulator-derived scores](02_build_cost_matrix_from_ie.ipynb), [coupled exact maps](06_coupled_policy_map.ipynb) and [scaling cases](07_scaling_policy_maps.ipynb). The [QUBO notebook](03_policy_problem_to_qubo.ipynb) introduces a binary energy for the same objective.

**Under development.** Coupling justified by physical transitions, robust objectives and larger structured classical baselines. See the [workflow](../docs/workflow.md).
