# System overview

> Can information, measurement and feedback be used to extract useful work from thermal fluctuations, and how can estimator/controller policies for such a system be synthesized and implemented?

A bead in an optical trap experiences thermal motion. The sensor measures position, the estimator infers state, and a classical controller moves the trap when its policy calls for a response. The resulting work depends on both the motion and the intervention. Measurement and memory are part of the physical accounting.

**Workflow.** Physics and information theory → model and measure → estimate and control → simulate fixed policies → build a local performance matrix → synthesize a policy map → replay and validate. FPGA/RFSoC is the intended fast-control platform. Quantum engineering is the current active focus within offline synthesis.

**Established.** A modular IE simulator, a fixed-policy pilot, exact small-instance optimization and ideal/noisy quantum simulations on synthetic targets.

**Work in progress.** Device-specific quantum benchmarking and stronger application baselines.

**Not yet implemented.** FPGA deployment, held-out policy-map replay, and real QPU execution in this repository.
