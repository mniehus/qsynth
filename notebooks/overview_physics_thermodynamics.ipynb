# Physics, thermodynamics and information

Thermal collisions make bead trajectories fluctuate. The optical trap confines the bead, and feedback changes the trap when the estimated state meets a trigger. Stochastic thermodynamics tracks work and heat along such trajectories [1]. Maxwell's demon motivates the accounting for information and the physical cost of resetting memory [2].

**Established in simulation.** An overdamped Langevin model, noisy measurement, trap motion and signed work accounting. See the [model and parameter units](../docs/information-engine.md) and [Notebook 02](02_build_cost_matrix_from_ie.ipynb).

**Under development.** Experimental calibration, a full energy budget, and direct tests of the fluctuation relations. The simulator score is not a measurement of apparatus efficiency.
