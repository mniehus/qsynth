# Quantum engineering for offline synthesis

Quantum engineering is the current active focus. Encode the policy-map objective in an operator, tune the QAOA circuit, and compare decoded samples with an exact classical baseline. A logical circuit becomes a device circuit only after placement, routing and native gate translation.

**Established for teaching instances.** [Bell and environment check](00_environment_and_bell.ipynb), [QUBO](03_policy_problem_to_qubo.ipynb), [Ising](04_qubo_to_ising.ipynb), [QAOA](05_qaoa_2x2.ipynb), [encodings](08_encoding_comparison.ipynb), [compilation](09_hardware_aware_transpilation.ipynb) and [noisy Aer](10_noisy_device_aware_aer.ipynb). QAOA alternates cost and mixer operations [1].

**Work in progress.** Benchmarking against stronger classical methods and obtaining a calibrated target. The existing hardware model is synthetic. Euro-Q-Exa is a prospective resource [2]. Project access and execution are not recorded.
