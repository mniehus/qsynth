# Control, estimation and real time

The estimator turns noisy sensor readings into a state estimate. A feedback policy maps that estimate to actuator commands. The fast loop stays classical. An FPGA/RFSoC implementation must meet the sampling deadline, including conversion, estimation, decisions and actuation.

**Established in simulation.** Separate blocks for plant, measurement, estimation, control and actuation. [Notebook 02](02_build_cost_matrix_from_ie.ipynb) bypasses adaptation to compare fixed policies.

**Not yet implemented.** Hardware deployment, fixed-point analysis and hardware-in-the-loop timing. See the [architecture](../docs/architecture.md) and [FPGA timing discussion](../docs/fpga.md).
